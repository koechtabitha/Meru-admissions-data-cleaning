```python
data_filtered = data[data['AdmissionNumber'].astype(str).str.startswith(('RI', 'FCME'), na=False)]

display(data_filtered.head())
```

```python
data = data[~data['AdmissionNumber'].str.startswith('FCMER', na=False)]
display(data.head())
```

```python
display(data_filtered.tail())
```

```python
data_filtered.loc[:, 'AdmissionNumber'] = data_filtered['AdmissionNumber'].astype(str).str.upper()
data_filtered.loc[:, 'ChildsName_First'] = data_filtered['ChildsName_First'].astype(str).str.upper()
data_filtered.loc[:, 'ChildsName_Last'] = data_filtered['ChildsName_Last'].astype(str).str.upper()

display(data_filtered[['AdmissionNumber', 'ChildsName_First', 'ChildsName_Last']].head())
```

```python
print(f"Shape before dropping duplicates: {data_filtered.shape}")
```

```python
print(data_filtered.columns.tolist())
```

```python
print(f"Shape before dropping duplicates: {data_filtered.shape}")

duplicate_subset_cols = [
    'AdmissionNumber',
    'ChildsName_First',
    'ChildsName_Last',
    'Residence_Line1',
    'Residence_City',
    'Phone'
]

# Drop duplicates based on the defined subset of columns
data_unique = data_filtered.drop_duplicates(subset=duplicate_subset_cols)

print(f"Shape after dropping duplicates: {data_unique.shape}")
display(data_unique[['AdmissionNumber', 'ChildsName_First', 'ChildsName_Last', 'Residence_City', 'Phone']].head())
```

```python
dropped_duplicates = data_filtered[data_filtered.duplicated(subset=duplicate_subset_cols, keep='first')]

print("Dropped duplicates (Child's Name, Admission Number, Residence City, Phone):")
display(dropped_duplicates[['ChildsName_First', 'ChildsName_Last', 'AdmissionNumber', 'Residence_City', 'Phone']])
```

```python
output_path = '/content/drive/MyDrive/FurahaAdmissionsUnique.csv'
data_unique.to_csv(output_path, index=False)
print(f"DataFrame saved to: {output_path}")
```
